```python
import pandas as pd 
import numpy as np

cta_delays = pd.read_csv("./../data/bus_delay_notices.csv")
cta_delays.head()
```

```python
cta_delays = cta_delays.assign(
    time_minutes=lambda x: pd.to_timedelta(x['Time Delayed'] + ":00").dt.total_seconds() / 60
)
cta_delays.head()

delays_by_route = cta_delays.groupby('Bus Route')['Time Delayed'].count()

delays_by_route.head()
```

```python
delays_by_route_sorted = delays_by_route.sort_values(ascending = False)
delays_by_route_sorted.head(10)
```
